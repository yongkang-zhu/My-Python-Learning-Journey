# Psychology Questionnaire Analyzer

## 1. 项目简介

这是一个基于 Python 的心理学问卷数据分析工具。
程序可以读取 CSV 格式的问卷数据，
完成数据检查、描述性统计和频数分析，
并自动导出分析结果。

## 2. 主要功能

- 读取 CSV 格式的问卷数据
- 统计样本数量和问卷题目
- 检查缺失值和无效答案
- 计算各题目的均值、标准差、最大值和最小值
- 统计各选项的频数和百分比
- 将分析结果导出为 CSV 文件

## 3. 使用技术

Python、csv、statistics、collections、os

## 4. 数据说明

本项目使用模拟问卷数据进行演示，
每道题采用 1-5 分的评分方式。

## 5. 运行方法

将 questionnaire_analyzer.ipynb
和 sample_questionnaire.csv 放在同一文件夹中，
使用 Jupyter Notebook 打开文件，
依次运行所有代码单元格即可。

## 6. 输出结果

程序会自动生成 output 文件夹，
其中包含描述性统计结果和频数分析结果。

```python
import os
import csv
import statistics
from collections import Counter

#查看当前工作目录
print("当前工作目录：", os.getcwd())

#检查问卷数据文件是否存在
file_path="sample_questionnaire.csv"

if os.path.exists(file_path):
    print("问卷数据文件已找到！")
else:
    print("没有找到数据文件，请检查文件位置。")
```

```python
#读取问卷数据
with open(file_path,"r",encoding="utf-8-sig")as file:
    reader=csv.DictReader(file)
    data=list(reader)

#查看数据基本情况
print("成功读取问卷数据！")
print("样本数量：",len(data))
print("问卷题目：",reader.fieldnames[1:])

#展示前5名被试的数据
print("\n前5名被试的数据：")
for participant in data[:5]:
    print(participant)
```

```python
#获取所有问卷题目
questions=reader.fieldnames[1:]

#记录数据检查结果
missing_count=0
invalid_count=0

#存储清洗后的数据
clean_data=[]

for participant in data:
    clean_participant={}

    for question in questions:
        answer=participant[question].strip()

        if answer=="":
            missing_count +=1
            clean_participant[question]=None

        elif answer.isdigit() and 1 <= int(answer) <=5:
                 clean_participant[question]=int(answer)

        else:
            invalid_count+=1
            clean_participant[question]=None
        
clean_data.append(clean_participant)

print("数据检查完成！")
print("总样本数：",len(data))
print("问卷题目数：",len(questions))
print("缺失答案数量：",missing_count)
print("无效答案数量：",invalid_count)
```

```python
#保存每道题的描述性统计结果
descriptive_results=[]

for question in questions:
    #提取当前题目所有有效答案
    scores=[
        participant[question]
        for participant in clean_data
        if participant[question] is not None
    ]

    #计算统计指标
    n=len(scores)

    if n>0:
        mean_score=statistics.mean(scores)
        min_score=min(scores)
        max_score=max(scores)

        #只有两个及以上有效答案时才能计算样本标准差
        if n>=2:
            sd_score=statistics.stdev(scores)
        else:
            sd_score=None
    else:
        mean_score=None
        sd_score=None
        min_score=None
        max_score=None

    #保存结果
descriptive_results.append({
        "题目":question,
        "有效样本数":n,
        "均值":round(mean_score,2)if mean_score is not None else None,
        "标准差":round(sd_score,2)if sd_score is not None else None,
        "最小值":min_score,
        "最大值":max_score
    })

 #展示统计结果
print("问卷描述性统计结果：\n")

for result in descriptive_results:
    print(result)
```

```python
#保存频数分析结果
frequency_results=[]
for question in questions:
    #提取当前题目的有效答案
    scores=[
        participant[question]
        for participant in clean_data
        if participant[question]is not None
    ]

    #统计各个分数出现的次数
    score_counts=Counter(scores)

    #计算总有效答案数量
    total=len(scores)

    #统计1-5分的频数和百分比
    for score in range(1,6):

        frequency=score_counts.get(score,0)

        if total>0:
            percentage=frequency/total*100
        else:
            percentage=0

        frequency_results.append({
               "题目":question,
               "选项":score,
               "频数":frequency,
               "百分比":round(percentage,2)
        })
#展示频数分析结果
print("问卷频数与百分比分析结果：\n")

for result in frequency_results:
    print(result)
```

```python
#创建输出文件夹
os.makedirs("output",exist_ok=True)

#导出描述性统计结果
with open(
    "output/descriptive_statistics.csv",
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:
    fieldnames=[
        "题目","有效样本数","均值",
        "标准差","最小值","最大值"
    ]

    writer=csv.DictWriter(file,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(descriptive_results)

#导出频数分析结果
with open(
    "output/frequency_analysis.csv",
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:

    fieldnames=["题目","选项","频数","百分比"]

    writer=csv.DictWriter(file,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(frequency_results)

print("分析结果已成功导出！")
print("文件位置：output文件夹")
```

```python
#输出最终分析报告
print("="*45)
print("   心理学问卷数据分析报告")
print("="*45)

print("样本总数：",len(data))
print("问卷题目数：",len(questions))
print("缺失答案数量：",missing_count)
print("无效答案数量：",invalid_count)

print("\n各题目有效样本数：")

for result in descriptive_results:
    print(
        result["题目"],
        "：",
        result["有效样本数"],
        "人"
    )

print("\n分析文件已保存：")
print("1. output/descriptive_statistics.csv")
print("2. output/frequency_analysis.csv")

print("=" * 45)
print("问卷数据分析完成！")
```
